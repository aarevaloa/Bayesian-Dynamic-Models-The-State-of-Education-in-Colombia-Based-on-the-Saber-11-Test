```python
# catalogo: (etiqueta, funcion) — cada funcion recibe (ruta_csv, dir_figuras, formatos)
CATALOGO = [
    ("Parte 1 - Familias estructurales",            figura_parte1),
    ("Parte 2 - Robustez estatica",                 figura_parte2),
    ("Parte 2 - Mini-figura S15 (contextual)",      minifigura_s15),
    ("Parte 3 - Dimension temporal y aditividad",   figura_parte3),
    ("Parte 3 - Mini-figura D07/D08",               minifigura_d07d08),
    ("Parte 4 - Limites del enfoque",               figura_parte4),
    ("Parte 5 - El WAIC como criterio de seleccion", figura_parte5),
]
```

```python
def aplanar_rutas(resultado):
    """Aplana cualquier combinación de listas, tuplas o strings devueltos por fu.guardar."""
    rutas = []
    if isinstance(resultado, (list, tuple)):
        for elem in resultado:
            rutas.extend(aplanar_rutas(elem))
    elif isinstance(resultado, str):
        rutas.append(resultado)
    return rutas


def generar_todas(
    ruta_csv=r"C:/Users/USUARIO/OneDrive - Universidad Nacional de Colombia/Tesis/Escritorio/Códigos/Simulaciones/resultados_estudio/tabla_resumen_2.csv",
    dir_figuras="figuras",
    formatos=("png", "pdf"),
):
    import os
    import time

    t0 = time.time()

    # Comprobar la existencia del CSV sin usar __file__
    if not os.path.exists(ruta_csv):
        try:
            base_dir = os.path.dirname(__file__)
        except NameError:
            base_dir = os.getcwd()  # Fallback seguro para Jupyter Notebooks

        alt = os.path.join(base_dir, os.path.basename(ruta_csv))
        if os.path.exists(alt):
            ruta_csv = alt
        else:
            print(f"[maestro] ERROR: no encuentro el CSV en: {ruta_csv}")
            return

    print(f"[maestro] CSV: {ruta_csv}")
    print(f"[maestro] Carpeta de salida: {os.path.abspath(dir_figuras)}")
    print(f"[maestro] Formatos: {', '.join(formatos)}\n")

    resumen = []
    for etiqueta, funcion in CATALOGO:
        try:
            resultado = funcion(
                ruta_csv, dir_figuras=dir_figuras, formatos=formatos
            )
            rutas_planas = aplanar_rutas(resultado)
            nombres_archivos = sorted(
                list(set(os.path.basename(r) for r in rutas_planas))
            )
            archivos = ", ".join(nombres_archivos)

            print(f"[maestro] OK   {etiqueta:48s} -> {archivos}")
            resumen.append((etiqueta, "OK"))
        except Exception as e:
            print(f"[maestro] FALLO {etiqueta:48s} -> {type(e).__name__}: {e}")
            resumen.append((etiqueta, f"ERROR: {e}"))

    dt = round(time.time() - t0, 1)
    n_ok = sum(1 for _, est in resumen if est == "OK")
    print("\n" + "=" * 70)
    print(
        f"[maestro] Generadas {n_ok} de {len(resumen)} ejecuciones en {dt} s."
    )
    print(f"[maestro] Carpeta de salida: {os.path.abspath(dir_figuras)}")
    incidencias = [(et, es) for et, es in resumen if es != "OK"]
    if incidencias:
        print("[maestro] Figuras con incidencias:")
        for et, es in incidencias:
            print(f"            {et}: {es}")
    print("=" * 70)
    return resumen
```

```python
if __name__ == "__main__":
    # ruta del CSV: primer argumento de linea de comandos, o el valor por defecto
    ruta = r"C:/Users/USUARIO/OneDrive - Universidad Nacional de Colombia/Tesis/Escritorio/Códigos/Simulaciones/resultados_estudio/tabla_resumen.csv"
    generar_todas(ruta)
```
